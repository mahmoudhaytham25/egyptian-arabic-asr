# 01 — YouTube Playlist Link Scraper

Pulls every video URL out of a YouTube playlist with `yt-dlp` (flat extraction — metadata only, no download). These URLs feed the dataset pipeline in `02_dataset_pipeline.ipynb`.

```python
import yt_dlp
```

```python
def get_playlist_details(playlist_url):
    # Options for yt-dlp to fetch playlist information without downloading videos
    ydl_opts = {
        'quiet': True,
        'extract_flat': True,  # Extract metadata without downloading videos
    }

    video_urls = []

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        playlist_info = ydl.extract_info(playlist_url, download=False)
        if 'entries' in playlist_info:  # Ensure it's a playlist
            for video in playlist_info['entries']:
                # Skip private videos
                if video.get('is_private', False):
                    print(f"Skipping private video: {video.get('title', 'Unknown Title')}")
                    continue
                video_url = video.get('url')
                if video_url and not video_url.startswith('http'):
                    video_url = f"https://www.youtube.com/watch?v={video_url}"
                print(video_url)
                video_urls.append(video_url)

    return video_urls
```

```python
playlist_url = "https://www.youtube.com/playlist?list=PL7TpI0K9I87Ip1qaWo6G_8dIcOe3k2jwz"
video_urls = get_playlist_details(playlist_url)

print(f"\nCollected {len(video_urls)} video URLs")
```
